## 데이터 프레임 생성

```python
import pandas as pd
import numpy as np

## requests로부터 response, body, item 항목 읽기
data = response.json()['response']['body']['items']['item']
df=pd.DataFrame(data)
df['exmn_dd_cnvs_prc'] = pd.to_numeric(df['exmn_dd_cnvs_prc'], errors='coerce')
df['exmn_dd_prc'] = pd.to_numeric(df['exmn_dd_prc'], errors='coerce')

df.info()
print(df.describe(include='all'))
print(df)

#df.describe(include=[np.number]) #수치형
#df.describe(include=['object']) #범주형
#df['exmn_dd_prc'].describe() #특정 열
#print(df.describe(include='all'))
df
```

```python

df['date']=pd.to_datetime(df['exmn_ymd'])
df['year']=df['date'].dt.year
df['month']=df['date'].dt.month
df['season']=df['month'].map(str).replace(
    {'1':'Winter','2':'Winter','3':'Spring','4':'Spring','5':'Spring','6':'Summer',
     '7':'Summer','8':'Summer','9':'Fall','10':'Fall','11':'Fall','12':'Winter'})

df.info()
print(df.describe(include='all'))
print(df)
```
